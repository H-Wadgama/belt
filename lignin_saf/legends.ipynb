```python
# Code that provides a legend for the breakdown
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
plt.rc('font',family='Arial')
# Define categories and corresponding colors from original vertical legend
categories_styled = [
    'Poplar feedstock', 
    'Utilities', 
    'Raw materials', 
    'Electricity', 
    'Fixed costs', 
    'Co-products', 
    'Capital depreciation', 
    'Average income tax',
    'Average ROI'
]

colors_styled = [
    '#9FC6C3',  # Installed costs
    "#DFECEB",  # Catalyst replacement
    "#EBE0E9",  # Utilities excl. electricity
    "#39605D",  # Process electricity
    '#C2A3BC',  # Fixed costs
    '#5FA09A',  # Hydrogen
    '#99668F',  # Ethanol
    '#772522',  # Co-product
    '#DD8B88',
]

patches = [mpatches.Patch(facecolor=color, label=label, edgecolor='black', linewidth=0) 
           for color, label in zip(colors_styled, categories_styled)]

fig, ax = plt.subplots(figsize=(10, 2))

# Create legend
legend = ax.legend(
    handles=patches,
    loc='center',
    ncol=2,
    frameon=False,        # No frame around the entire legend
    fontsize=16,
    handlelength=2,
    handletextpad=0.3,
    borderpad=10.8,
    columnspacing=0.8,
    labelspacing=0.3
)

# Ensure no text boxes
for text in legend.get_texts():
    text.set_bbox(None)

plt.rcParams['svg.fonttype'] = 'none'

ax.axis('off')
plt.tight_layout()
plt.savefig("msp_breakdown_legend.svg")
plt.show()
```
